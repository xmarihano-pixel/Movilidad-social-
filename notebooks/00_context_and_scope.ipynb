# 00 · Contexto, alcance y preguntas (Fases 0, 1 y 2)

**Proyecto:** ¿Qué tan pegajosa es la pobreza? Movilidad social en México.

## Pregunta central
¿En qué medida las condiciones de origen se asocian con la posición económica adulta en México, y qué circunstancias aumentan o reducen la probabilidad de **seguir en el 20% más pobre**?

## Preguntas específicas
| # | Pregunta | Cómo se responde |
|---|---|---|
| P1 | ¿Qué tanto se hereda la posición socioeconómica? | Matriz de transición de quintiles (notebook 03) |
| P2 | De quienes nacen en el 20% más pobre, ¿cuántos siguen ahí? | `sigue_q1` ponderado (03) |
| P3 | ¿Qué circunstancias de origen se asocian con quedarse? | Regresión logística y modelos de ML (05, 06) |
| P4 | ¿Las brechas son iguales por región, sexo, lengua y tono de piel? | Análisis por grupos y desempeño por subgrupo (03, 08) |
| P5 | ¿El contexto del estado importa? | Comparación con pobreza CONEVAL e informalidad ENOE (03) |

## Hipótesis (se registran **antes** de ver resultados)
- **H1.** Más de 40% de quienes nacen en Q1 siguen en Q1.
- **H2.** Crecer en el Sur o en una localidad rural aumenta la probabilidad de seguir en Q1.
- **H3.** La escolaridad de la madre se asocia con menor persistencia.
- **H4.** Hablar lengua indígena y tener tono de piel más oscuro se asocian con mayor persistencia, aun comparando personas con origen similar.
- **H5.** Los modelos complejos no superan por mucho a una regresión logística.

## Marco conceptual (Fase 2)
- **Movilidad intergeneracional relativa:** comparar el lugar (quintil) que ocupa una persona entre los de su generación hoy, con el lugar que ocupaba su hogar cuando tenía 14 años.
- **Posición socioeconómica:** índice de bienes y servicios del hogar (no ingreso), porque la EMOVI no pregunta el ingreso de los padres. Es la práctica del CEEY.
- **Pobreza "pegajosa":** nacer en el quintil 1 y seguir en el quintil 1 de adulto.
- **Advertencia:** todo es **asociación**. La EMOVI es una encuesta transversal con preguntas retrospectivas; no permite afirmar causas.

## Decisiones congeladas antes de ver resultados
1. Variable objetivo principal: `sigue_q1` (sí/no) entre quienes nacieron en Q1.
2. Variable objetivo secundaria: `movilidad` (Ascenso / Permanencia / Descenso).
3. Quintiles ponderados con el factor de expansión y calculados **dentro de cada cohorte de edad**.
4. Predictores: sólo circunstancias de origen o dadas. La escolaridad propia es **mediadora** y va en un modelo aparte.
5. Regla de selección: si la regresión logística está a ≤ 0.01 de AUC del mejor modelo, se elige la logística.
